```python
df_spark = spark.read.text(
    '/Volumes/workspace/testschema/testunitycolume/pg1513.txt'
)
display(df_spark)
```

```python
from pyspark.sql.functions import explode, split, col

# Mapper: Split lines into words
words_df = df_spark.select(
    explode(
        split(col("value"), " ")
    ).alias("word")
)

# Reducer: Group by word and count occurrences
word_counts = words_df.groupBy("word").count()

display(word_counts)
```
